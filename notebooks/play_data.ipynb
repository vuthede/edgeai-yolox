```python
import json 

root = "/home/vuthede/fiftyone/coco-2017/validation"
label = f'{root}/annotations/person_keypoints_val2017.json'

data = json.load(open(label))
print(data.keys())
data['annotations'][0].keys()

# For each annotation add 3 fields (weight, height, age)  with random values. if this annotation do not have keypoints, then add -1 to weight, height, age
# Save the new json file to a new file called person_keypoints_val2017_with_random_bimometry.json
import random
for annotation in data['annotations']:
    if len(annotation['keypoints']) == 0:
        annotation['weight'] = -1
        annotation['height'] = -1
        annotation['age'] = -1
    else:
        annotation['weight'] = random.randint(40, 100)
        annotation['height'] = random.randint(140, 200)
        annotation['age'] = random.randint(18, 80)

json.dump(data, open(f'{root}/annotations/person_keypoints_val2017_with_random_bimometry.json', 'w'))
print('Done')
```
